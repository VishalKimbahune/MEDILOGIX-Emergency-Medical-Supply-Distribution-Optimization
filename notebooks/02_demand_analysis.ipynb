# 02 — MEDILOGIX Supply vs. Demand Gap Analysis

Analyze product-wise supply coverage, priority-tier demand distribution, and controlled emergency shortages.

```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path('../data/synthetic')
products = pd.read_csv(DATA_DIR / 'products.csv')
inventory = pd.read_csv(DATA_DIR / 'inventory.csv')
demands = pd.read_csv(DATA_DIR / 'demands.csv')

supply_agg = inventory.groupby('product_id')['quantity_available'].sum().rename('total_supply')
demand_agg = demands.groupby('product_id')['quantity_requested'].sum().rename('total_demand')
balance = products[['product_id', 'name', 'unit']].merge(supply_agg, on='product_id').merge(demand_agg, on='product_id')
balance['net_balance'] = balance['total_supply'] - balance['total_demand']
balance
```
